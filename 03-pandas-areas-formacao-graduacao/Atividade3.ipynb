# Aula 3 — Áreas gerais de formação na graduação (IBGE)

**Tabela 3** — Pessoas com nível superior completo, por áreas gerais de formação do curso de graduação concluído, por sexo, segundo Brasil, Grandes Regiões e Unidades da Federação.
Fonte: IBGE — Censo Demográfico.

**Objetivo:** dividir a tabela em 3, com homens e mulheres por região:
1. com curso superior (total)
2. com curso nas áreas de **Ciência, Tecnologia, Engenharias e Matemática**
3. com curso nas áreas de **Educação, Serviços pessoais, Saúde e Bem-estar**

Depois salvar cada uma em `.xlsx` e apresentar as estatísticas com `describe()`.

```python
# No terminal, uma vez: python -m pip install pandas openpyxl
import pandas as pd
```

```python
# arquivo na mesma pasta do notebook
path = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

## Olhando o arquivo bruto

Assim como na aula anterior, o arquivo do IBGE tem **título**, cabeçalho em **duas linhas** e **fonte no rodapé**.

```python
bruto = pd.read_excel(path, header=None)
bruto
```

## Limpando a tabela

- Linhas 0–2: título e cabeçalho  |  3–35: dados  |  36: fonte
- Cada área tem 3 colunas: Total, Homens, Mulheres

```python
tabela_3 = bruto.iloc[3:36].copy()
tabela_3.columns = [
    "regiao",
    "superior_total", "superior_homens", "superior_mulheres",
    "ctem_total", "ctem_homens", "ctem_mulheres",
    "educacao_saude_total", "educacao_saude_homens", "educacao_saude_mulheres",
]
tabela_3 = tabela_3.reset_index(drop=True)

# garante que os números são numéricos
for col in tabela_3.columns[1:]:
    tabela_3[col] = pd.to_numeric(tabela_3[col], errors="coerce").astype(int)

tabela_3
```

## Filtrando só as Grandes Regiões

O pedido é **por região**, então ficamos só com Norte, Nordeste, Sudeste, Sul e Centro-Oeste.
A linha *Brasil* fica de fora porque é a soma das regiões e distorceria as estatísticas do `describe()`.

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

por_regiao = tabela_3[tabela_3["regiao"].isin(regioes)].reset_index(drop=True)
por_regiao
```

## Tabela 1 — Homens e mulheres por região com curso superior

```python
curso_superior = por_regiao[["regiao", "superior_homens", "superior_mulheres"]].rename(
    columns={"superior_homens": "homens", "superior_mulheres": "mulheres"}
)
curso_superior
```

## Tabela 2 — Homens e mulheres por região com curso em Ciência, Tecnologia, Engenharias e Matemática

```python
tecnologia = por_regiao[["regiao", "ctem_homens", "ctem_mulheres"]].rename(
    columns={"ctem_homens": "homens", "ctem_mulheres": "mulheres"}
)
tecnologia
```

## Tabela 3 — Homens e mulheres por região com curso em Educação, Serviços pessoais, Saúde e Bem-estar

```python
educacao_saude = por_regiao[["regiao", "educacao_saude_homens", "educacao_saude_mulheres"]].rename(
    columns={"educacao_saude_homens": "homens", "educacao_saude_mulheres": "mulheres"}
)
educacao_saude
```

## Salvando as 3 tabelas em `.xlsx`

Os arquivos são salvos na mesma pasta do notebook.

```python
curso_superior.to_excel("curso_superior.xlsx", index=False)
tecnologia.to_excel("tecnologia.xlsx", index=False)
educacao_saude.to_excel("educacao_saude.xlsx", index=False)

print("Arquivos salvos: curso_superior.xlsx, tecnologia.xlsx, educacao_saude.xlsx")
```

## Estatísticas com `describe()`

O `describe()` mostra, para cada coluna numérica:
- `count` → quantidade de valores (5 regiões)
- `mean` → média
- `std` → desvio padrão (o quanto os valores variam)
- `min` / `max` → menor e maior valor
- `25%`, `50%`, `75%` → quartis (o `50%` é a mediana)

### Curso superior

```python
curso_superior.describe().round(2)
```

### Ciência, Tecnologia, Engenharias e Matemática

```python
tecnologia.describe().round(2)
```

### Educação, Serviços pessoais, Saúde e Bem-estar

```python
educacao_saude.describe().round(2)
```

## Conclusão

```python
tabelas = {
    "Curso superior": curso_superior,
    "Ciência, Tecnologia, Eng. e Matemática": tecnologia,
    "Educação, Serv. pessoais, Saúde e Bem-estar": educacao_saude,
}

for nome, t in tabelas.items():
    h, m = t["homens"].sum(), t["mulheres"].sum()
    print(f"{nome}")
    print(f"  Média por região -> homens: {t['homens'].mean():,.0f} | mulheres: {t['mulheres'].mean():,.0f}")
    print(f"  Mulheres são {m / (h + m) * 100:.1f}% do total\n")
```
