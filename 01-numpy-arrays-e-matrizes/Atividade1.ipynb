# Atividade 1 — NumPy

Exercícios de criação de arrays e matrizes com NumPy, usando `ndim`, `shape` e `mean`.

```python
import numpy as np
```

## 1 - Crie um array com 20 elementos e uma dimensão. Printe ele.

```python
array = np.arange(1, 21)
print(array)
```

## 2 - Faça a média desse array.

```python
media = array.mean()
print("Média do array:", media)
```

## 3 - Crie uma matriz com 10 colunas e 20 linhas.

```python
matriz = np.random.randint(0, 100, size=(20, 10))
print(matriz)
```

## 4 - Qual o shape da matriz?

```python
print("Shape da matriz:", matriz.shape)
```

A matriz tem **20 linhas** e **10 colunas**, por isso o shape é `(20, 10)`.

## 5 - Utilize os comandos ndim, shape, mean.

```python
print("=== Array ===")
print("ndim :", array.ndim)
print("shape:", array.shape)
print("mean :", array.mean())

print()
print("=== Matriz ===")
print("ndim :", matriz.ndim)
print("shape:", matriz.shape)
print("mean :", matriz.mean())
```

- `ndim` → número de dimensões (o array tem 1, a matriz tem 2)
- `shape` → tamanho de cada dimensão
- `mean` → média de todos os elementos
